```python
# Task 24: Data Type Cleaning

## Objective
Identify incorrect data types in a dataset and convert
columns into appropriate data types.

## Deliverables
- Original data types
- Corrected data types
- Data type comparison
- Explanation of each conversion
- Corrected dataset
```

```python
import pandas as pd
import numpy as np

print("Libraries imported successfully!")
```

```python
from pathlib import Path

csv_content = """employee_id,employee_name,age,salary,department,joining_date,performance_score
E001,Aarav,25,"₹45,000",IT,2023-01-15,4.5
E002,Diya,28,"₹52,000",HR,2022-07-10,4.2
E003,Kabir,31,"₹68,500",Finance,2021-03-22,4.7
E004,Ananya,26,"₹49,000",IT,2023-09-05,4.4
E005,Ishaan,35,"₹75,000",Operations,2020-11-18,4.1
E006,Meera,29,"₹58,000",HR,2022-02-14,4.6
E007,Rohan,33,"₹72,500",Finance,2021-08-30,4.3
E008,Sara,24,"₹42,000",IT,2024-01-08,4.8
E009,Vivaan,30,"₹63,000",Operations,2022-05-16,4.0
E010,Aditi,27,"₹55,500",HR,2023-04-12,4.5
"""

Path("employee_data_raw.csv").write_text(
    csv_content,
    encoding="utf-8"
)

print("Original employee dataset created successfully!")
```

```python
df = pd.read_csv(
    "employee_data_raw.csv",
    dtype="string"
)

print("Dataset loaded successfully!")

display(df)
```

```python
print("ORIGINAL DATA TYPES")

display(df.dtypes.to_frame(
    name="Original Data Type"
))
```

```python
print("ORIGINAL DATASET INFORMATION")

df.info()
```

```python
cleaned_df = df.copy()

print("Created a copy for data type conversion.")
```

```python
cleaned_df["age"] = (
    cleaned_df["age"].astype("int64")
)

print("Age converted into integer successfully.")

display(cleaned_df[["employee_name", "age"]])
```

```python
cleaned_df["salary"] = pd.to_numeric(
    cleaned_df["salary"]
    .str.replace("₹", "", regex=False)
    .str.replace(",", "", regex=False),
    errors="raise"
)

print("Salary converted into numerical format.")

display(cleaned_df[["employee_name", "salary"]])
```

```python
cleaned_df["performance_score"] = pd.to_numeric(
    cleaned_df["performance_score"],
    errors="raise"
).astype("float64")

print("Performance scores converted into float.")

display(
    cleaned_df[["employee_name", "performance_score"]]
)
```

```python
cleaned_df["joining_date"] = pd.to_datetime(
    cleaned_df["joining_date"],
    format="%Y-%m-%d",
    errors="raise"
)

print("Joining dates converted into datetime.")

display(
    cleaned_df[["employee_name", "joining_date"]]
)
```

```python
cleaned_df["department"] = (
    cleaned_df["department"].astype("category")
)

print("Department converted into categorical format.")

display(
    cleaned_df[["employee_name", "department"]]
)
```

```python
print("IDENTIFIER AND NAME DATA TYPES")

print("Employee ID:", cleaned_df["employee_id"].dtype)
print("Employee Name:", cleaned_df["employee_name"].dtype)
```

```python
conversion_notes = {
    "employee_id": "Kept as string: unique identifier",
    "employee_name": "Kept as string: text information",
    "age": "String to int64 using astype()",
    "salary": "Currency text to float64 using pd.to_numeric()",
    "department": "String to category using astype()",
    "joining_date": "String to datetime using pd.to_datetime()",
    "performance_score": "String to float64 using pd.to_numeric()"
}

dtype_comparison = pd.DataFrame({
    "Original Data Type": [
        str(df[column].dtype)
        for column in df.columns
    ],

    "Corrected Data Type": [
        str(cleaned_df[column].dtype)
        for column in df.columns
    ],

    "Conversion Explanation": [
        conversion_notes[column]
        for column in df.columns
    ]
}, index=df.columns)

dtype_comparison.index.name = "Column"

print("ORIGINAL VS CORRECTED DATA TYPES")

display(dtype_comparison)
```

```python
print("CORRECTED DATA TYPES")

display(
    cleaned_df.dtypes.to_frame(
        name="Corrected Data Type"
    )
)

print("\nCORRECTED DATASET INFORMATION")

cleaned_df.info()
```

```python
print("FINAL CLEANED EMPLOYEE DATASET")

display(cleaned_df)
```

```python
from pandas.api.types import (
    is_integer_dtype,
    is_numeric_dtype,
    is_datetime64_any_dtype
)

assert is_integer_dtype(cleaned_df["age"])
assert is_numeric_dtype(cleaned_df["salary"])
assert is_numeric_dtype(cleaned_df["performance_score"])
assert is_datetime64_any_dtype(cleaned_df["joining_date"])

print("DATA TYPE VALIDATION")
print("--------------------")
print("Age is integer:", is_integer_dtype(cleaned_df["age"]))
print("Salary is numeric:", is_numeric_dtype(cleaned_df["salary"]))
print("Performance score is numeric:",
      is_numeric_dtype(cleaned_df["performance_score"]))
print("Joining date is datetime:",
      is_datetime64_any_dtype(cleaned_df["joining_date"]))
print("Department is categorical:",
      isinstance(cleaned_df["department"].dtype,
                 pd.CategoricalDtype))

print("\nAll data type conversions verified successfully!")
```

```python
print("EMPLOYEE SALARY ANALYSIS")

print("Average salary:", cleaned_df["salary"].mean())
print("Highest salary:", cleaned_df["salary"].max())
print("Lowest salary:", cleaned_df["salary"].min())

print("\nAVERAGE PERFORMANCE SCORE")

print(cleaned_df["performance_score"].mean())
```

```python
# Export corrected employee dataset
cleaned_df.to_csv(
    "employee_data_cleaned.csv",
    index=False
)

# Export data type comparison
dtype_comparison.to_csv(
    "data_type_comparison.csv"
)

print("All output files exported successfully!")
```
